Read the CSV file

```python
import pandas as pd

df = pd.read_csv('../data/auto.csv', index_col='ID')
df.head()
```

Count the number of observations

```python
df.count()
```

Drop duplicates

```python
df = df.drop_duplicates(
    subset=['CarNumber', 'Make_n_model', 'Fines'],
    keep='last'
)
df.head()
```

```python
df.count()
```

Check how many values are missing from each column.


```python
df.isna().sum()
```

Drop columns

```python

df['Refund'].mean()
```

```python
df = df.dropna(axis=1, thresh=df.shape[0] - 500)
df.isna().sum()
```

Fill missing values in Refund

```python
df['Refund'] = df['Refund'].fillna(method='ffill')
df.isna().sum()
```

Fill missing values in Fines
Replace missing values with the mean value of the column.

```python
df['Fines'] = df['Fines'].fillna(df['Fines'].mean())
df.isna().sum()
```

Split and parse Make_n_model


```python
df['Make'] = df['Make_n_model'].apply(lambda x: x.split()[0])
df['Model'] = df['Make_n_model'].apply(lambda x: ' '.join(x.split()[1:]))
df.head()
```

Drop Make_n_model

```python
df = df.drop(columns=['Make_n_model'])
df.head()
```

Save the dataframe to JSON

```python
df.to_json('auto.json', orient='records')
```
