## Read CSV

```python
import pandas as pd
df=pd.read_csv(
    '../data/feed-views.log',
    sep='\t',
    skiprows=[2,3],
    skipfooter=2,
    names=['datetime','user'],
    engine='python'
)
df=df.set_index('datetime')
df.index.name='date_time'
df.head()

```

## Save to CSV
