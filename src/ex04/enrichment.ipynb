```python
import pandas as pd
import numpy as np
import requests
```

Read the JSON file from ex02

```python
pd.options.display.float_format = '{:.2f}'.format
df = pd.read_json('../ex02/auto.json')
df.head()
```

Create a sample of 200 new observations

```python
sample = df.sample(n=200, replace=True, random_state=21)
sample.head()
```

Concatenate the sample with the initial dataframe

```python
concat_rows = pd.concat([df, sample], ignore_index=True)
concat_rows.tail()
```

Enrich the concat_rows dataframe with a new column containing gen

```python
np.random.seed(21)

year = pd.Series(
    np.random.randint(1980, 2020, size=len(concat_rows)),
    name='Year'
)

fines = pd.concat([concat_rows, year], axis=1)

fines.head()
```

```python
surnames = pd.read_json('../data/surname.json')
surnames.head()
```

```python
surnames.columns = surnames.iloc[0]
surnames = surnames.iloc[1:].reset_index(drop=True)
surnames.head()
```

Create a series with surnames

```python
surname_series = surnames['NAME'].str.replace(r'[^A-Z]', '', regex=True)
surname_series.head()
```

Unique car numbers

```python
unique_car_numbers = pd.Series(sample['CarNumber'].unique(), name='CarNumber')
unique_car_numbers.head()
```

Sample surnames for owners

```python
surname_sample = surname_series.sample(
    n=len(unique_car_numbers),
    random_state=21,
    replace=True
).reset_index(drop=True)

surname_sample.head()
```

Create the owners dataframe

```python
owners = pd.DataFrame({
    'CarNumber': unique_car_numbers.reset_index(drop=True),
    'SURNAME': surname_sample
})

owners.head()
```

Append five new observations to the fines dataframe

```python
new_fines = pd.DataFrame({
    'CarNumber': ['A001AA199', 'B002BB199', 'C003CC199', 'D004DD199', 'E005EE199'],
    'Refund': [np.nan, 1, 0, np.nan, 1],
    'Fines': [1200.00, 2300.00, 750.00, 980.00, 1500.00],
    'Make': ['Lada', 'BMW', 'Audi', 'Toyota', 'Ford'],
    'Model': ['Vesta', 'X5', 'A4', 'Camry', 'Focus'],
    'Year': [2010, 2012, 2015, 2018, 2011]
})

new_fines
```

```python
fines = pd.concat([fines, new_fines], ignore_index=True)
fines.tail()
```

```python
owners = owners.iloc[:-20]
owners.tail()
```

Add three new observations to owners

```python
new_owners = pd.DataFrame({
    'CarNumber': ['X111XX199', 'Y222YY199', 'Z333ZZ199'],
    'SURNAME': ['Adams', 'Clark', 'Lewis']
})

new_owners
```

```python
owners = pd.concat([owners, new_owners], ignore_index=True)
owners.tail()
```

```python
inner_join = fines.merge(owners, on='CarNumber', how='inner')
inner_join.info()
```

```python
outer_join = fines.merge(owners, on='CarNumber', how='outer')
outer_join.info()
```

Left join
This join keeps only the car numbers from the `fines` dataframe.

```python
left_join = fines.merge(owners, on='CarNumber', how='left')
left_join.head()
```

Right join
This join keeps only the car numbers from the `owners` dataframe.

```python
right_join = fines.merge(owners, on='CarNumber', how='right')
right_join.info()
```
