Create а dataframe

```python
import pandas as pd

views = pd.read_csv(
    '../data/feed-views.log',
    sep='\t',
    names=['datetime', 'user']
)

views.head()
```

Convert datetime type

```python
views['datetime']=pd.to_datetime(views['datetime'])
views.dtypes
```

Extract date and time parts

```python
views['year'] = views['datetime'].dt.year
views['month'] = views['datetime'].dt.month
views['day'] = views['datetime'].dt.day
views['hour'] = views['datetime'].dt.hour
views['minute'] = views['datetime'].dt.minute
views['second'] = views['datetime'].dt.second

views.head()
```

Create the daytime column

```python
bins = [-1, 3, 6, 10, 16, 19, 23]

labels = [
    'night',
    'early morning',
    'morning',
    'afternoon',
    'early evening',
    'evening'
]

views['daytime'] = pd.cut(
    views['hour'],
    bins=bins,
    labels=labels
)

views[['hour', 'daytime']].head()
```

Set user as the index

```python
views=views.set_index('user')
views.head()
```

Count the number of elements

```python
views.count()
```

value_counts

```python
views['daytime'].value_counts()
```

Sort the values by hour, minute, and second

```python
views=views.sort_values(by=['hour', 'minute', 'second'])
views.head(8)
```

Calculate the minimum and maximum hours and the mode for daytime

```python
min_hour = views['hour'].min()
max_hour = views['hour'].max()
mode_daytime = views['daytime'].mode()

min_hour, max_hour, mode_daytime
```

Calculate the maximum hour for the rows where time for days is morning

```python
min_morning_hour = views[views['daytime'] == 'morning']['hour'].min()

min_morning_hour
```

Calculate the maximum hour for night rows

```python
max_night_hour = views[views['daytime'] == 'night']['hour'].max()

max_night_hour
```

Find who visited the page during those hours

```python
night_example = views[
    (views['daytime'] == 'night') &
    (views['hour'] == max_night_hour)
].head(1)

night_example
```

```python
morning_example = views[
    (views['daytime'] == 'morning') &
    (views['hour'] == min_morning_hour)
].head(1)

morning_example
```

Calculate the mode for hour and daytime

```python
mode_hour = views['hour'].mode()
mode_daytime = views['daytime'].mode()

mode_hour, mode_daytime
```

Show the three earliest hours with corresponding usernames

```python
views.nsmallest(3, 'hour')
```

Show the three latest hours with corresponding usernames

```python
views.nlargest(3,'hour')
```

Find the most popular visiting interval

```python
desc=views.describe()

iqr=desc.loc['75%','hour']-desc.loc['25%','hour']

iqr, desc
```
