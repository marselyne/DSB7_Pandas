Load and read JSON

```python
import pandas as pd
df=pd.read_json('../ex02/auto.json')
df=df.set_index('CarNumber')
df.head()
```

Rows where fines are greater than 2100

```python
df[df['Fines']>2100]
```

 Rows where fines are greater than 2100 and refund equals 2

```python
df[(df['Fines']>2100)&(df['Refund']==2)]
```

Rows where the model is from the list
Focus and Corolla

```python
df[df['Model'].isin(['Focus', 'Corolla'])]
```

Rows where the car number is from the list

```python
df.loc[
    ['Y7689C197RUS', '92928M178RUS', '7788KT197RUS', 'H115YO163RUS', 'X758HY197RUS']
]
```

Median fines grouped by make

```python
df.groupby('Make')['Fines'].median()
```

Median fines grouped by make and model

```python
df.groupby(['Make', 'Model'])['Fines'].median()
```

Minimum and maximum fines grouped by make and model

```python
df.groupby(['Make', 'Model'])['Fines'].agg(['min', 'max'])
```

Display the standard deviation of fines grouped by make and model to better understand the variance.

```python
df.groupby(['Make', 'Model'])['Fines'].std()
```

Display the car numbers, grouped by the number of fines in descending order. We want to find those who most often violated the law.

```python
df.groupby(['CarNumber'])['Fines'].agg('count').sort_values(ascending=False)
```

Top-1 Fines Cars

```python
df.loc['Y7689C197RUS']
```

Most paid

```python
df.groupby(['CarNumber'])['Fines'].agg('sum').sort_values(ascending=False)
```

top-1 for most paid

```python
df.loc['X758HY197RUS']
```

Are there any car numbers connected to different models?

```python
df.groupby('CarNumber')['Model'].nunique().sort_values(ascending=False)
```
